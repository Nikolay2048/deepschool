### LinearRegression

```python
import torch
from torch import nn


class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.linear(x)


x = torch.tensor([
    [0.0],
    [1.0],
    [2.0],
    [3.0],
])

y = torch.tensor([
    [1.0],
    [3.0],
    [5.0],
    [7.0],
])

model = Model()
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(
    params=model.parameters(),
    lr=0.05,
)

model.train()

for epoch in range(100):
    prediction = model(x)
    loss = loss_fn(prediction, y)

    loss.backward()
    optimizer.step()
    optimizer.zero_grad()

    if epoch % 10 == 0:
        print(
            f"epoch={epoch:3d}, "
            f"loss={loss.item():.6f}"
        )
```

```python
print(model.linear.weight)
print(model.linear.bias)
```
